```python
from ucimlrepo import fetch_ucirepo

# Load features and the class column.
iris = fetch_ucirepo(id=53)
X = iris.data.features
y = iris.data.targets["class"]
```

```python
from sklearn.model_selection import train_test_split

# 80% training, 20% testing split
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
X_train.head(5)
```

```python
from time import perf_counter
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

# Standardize then perform logic regression to predict Y
pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegression(max_iter=1000))
])

# Measure training time.
start = perf_counter()
pipeline.fit(X_train, y_train)
training_time = perf_counter() - start\

print(f"Training complete. Training time: {training_time:.3f} seconds")
```

```python
from sklearn.metrics import accuracy_score

# Measure testing/prediction time.
start = perf_counter()
y_pred = pipeline.predict(X_test)
testing_time = perf_counter() - start
accuracy = accuracy_score(y_test, y_pred)

print(f"Prediction time: {testing_time:.6f} seconds")
print(f"Test accuracy: {accuracy:.2%}")
```

### Question 6: report
- The training time ranges from 0.005 to 0.006 seconds.
- The testing/prediction time is just over 0.001 second
- The test accuracy on the test data is 93.33%
